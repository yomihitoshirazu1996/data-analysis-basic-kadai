```python
import numpy as np
import matplotlib.pyplot as plt

fig, ax = plt.subplots(2,3)

y0 = [1, 2, -5, 2]
ax[0,0].plot(y0)
ax[0,0].set_title('折れ線グラフ')

x_sin = np.linspace(0, 10, 100)
y_sin = 2 + 2 * np.sin(2 * x_sin)
ax[0,1].plot(x_sin, y_sin)
ax[0,1].set_title('Sin関数')


x_hist = np.random.normal(15, 5, 2000)
ax[0,2].hist(x_hist, bins=50)
ax[0,2].set_title('ヒストグラム')


np.random.seed(3)
x_scatter = 4 + np.random.normal(0, 2, 60)
y_scatter = 4 + np.random.normal(0, 2, len(x_scatter))
ax[1,0].scatter(x_scatter, y_scatter)
ax[1,0].set_title('散布図')


plt.tight_layout()
plt.show()
```
