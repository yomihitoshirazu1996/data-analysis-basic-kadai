```python
def tax(x):
    return x * 1.10
```

```python

import pandas as pd

df = pd.read_csv("/content/sample_data/sample_pandas_6.csv")
tax_series = df['単価'].apply(tax)
tax_series.name = "単価（税込み）"
pd.concat([df, tax_series], axis=1)
```

```python
!pip install japanize-matplotlib
import matplotlib.pyplot as plt
import japanize_matplotlib
import pandas as pd

df = pd.read_csv("/content/sample_data/sample_pandas_6.csv")
category_df = pd.read_csv('/content/sample_data/category.csv')

merged_df = pd.merge(df, category_df, on='商品番号', how='left')

category_counts = merged_df['カテゴリー'].value_counts()

plt.figure(figsize=(10, 6))
category_counts.plot(kind='bar')
plt.title('カテゴリー別頻度')
plt.xlabel('カテゴリー')
plt.ylabel('頻度')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
